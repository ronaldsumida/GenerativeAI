# Understanding neural networks

In this lab, students are challenged to reverse-engineer a neural network given a set of weights and biases. Here is some code you can use to simulate the network and test the network with inputs and outputs.

```python
# Weights
wac = -0.82235616
wad = 1.1054902
wae = -0.6134991
waf = -0.59988433

wbc = 0.36023167
wbd = -1.1052716
wbe = 0.61324805
wbf = -0.42968866

wcg = 0.000111482186
wdg = 0.904603839
weg = -1.62997365
wfg = -0.0000618918712

# Biases
bc = 1.6475359
bd = 1.4028503
be = -0.7791468
bf = -0.42655456
bg = 0.7288665

def relu(x):
    return max(0, x)

def run(a, b):
    c = (a * wac) + (b * wbc) + bc
    d = (a * wad) + (b * wbd) + bd
    e = (a * wae) + (b * wbe) + be
    f = (a * waf) + (b * wbf) + bf
    g = (relu(c) * wcg) + (relu(d) * wdg) + (relu(e) * weg) + (relu(f) * wfg) + bg
    return g
```

And here are sample inputs and outputs:

```python
print(run(2, 2))
print(run(4, 2))
print(run(5, 0))
print(run(10, 5))
print(run(4, 8))
```
